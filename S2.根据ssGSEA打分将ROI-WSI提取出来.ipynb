label文件的准备：
1代表高风险组，0代表低风险组



```python
import os
import shutil
import pandas as pd

# 设置路径
source_folder = r'F:\WSI.RAW\TCGA.COAD'
target_folder = r'F:\TCGA.COAD.ROI'
csv_file = r'G:\1.Sprouts\CRC.Pathology\20.WSL\label.csv'  # 请根据实际位置修改路径

# 创建目标文件夹（如果不存在）
os.makedirs(target_folder, exist_ok=True)

# 读取ROI.csv文件
df = pd.read_csv(csv_file)
roi_ids = set(df['ID'].str.strip())  # 去除可能的空格

print(f'共有 {len(roi_ids)} 个感兴趣的样本')

# 记录状态
found_files = []
missing_files = []

# 遍历ROI列表
for sample_id in roi_ids:
    # 构建源文件路径（假设已经重命名为简化格式）
    source_file = os.path.join(source_folder, f'{sample_id}.svs')
    
    # 检查文件是否存在
    if os.path.exists(source_file):
        # 构建目标文件路径
        target_file = os.path.join(target_folder, f'{sample_id}.svs')
        
        # 复制文件（使用copy2保留元数据）
        shutil.copy2(source_file, target_file)
        found_files.append(sample_id)
        print(f'✓ 已转移: {sample_id}.svs')
    else:
        missing_files.append(sample_id)
        print(f'✗ 未找到: {sample_id}.svs')

# 输出统计信息
print('\n' + '='*50)
print(f'转移完成！')
print(f'成功转移: {len(found_files)} 个文件')
print(f'未找到: {len(missing_files)} 个文件')

# 将缺失的样本保存到文件
if missing_files:
    print('\n缺失的样本名:')
    with open(os.path.join(target_folder, 'missing_samples.txt'), 'w') as f:
        for sample in missing_files:
            print(f'  - {sample}')
            f.write(f'{sample}\n')
    print(f'\n缺失样本列表已保存到: {os.path.join(target_folder, "missing_samples.txt")}')
```
