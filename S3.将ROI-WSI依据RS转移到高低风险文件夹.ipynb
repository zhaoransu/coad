```python
import os
import shutil
import pandas as pd

def move_svs_files():
    # 1. 定义路径 (使用绝对路径)
    # 源文件目录
    source_dir = r"F:\TCGA.COAD.ROI"
    # CSV文件路径
    #csv_path = os.path.join(source_dir, "label.csv")
    csv_path = r'G:\1.Sprouts\CRC.Pathology\20.WSL\label.csv'
    # 目标目录
    low_dir = r"G:\1.Sprouts\CRC.Pathology\20.WSL\DataSet\Low"
    high_dir = r"G:\1.Sprouts\CRC.Pathology\20.WSL\Dataset\High"

    # 2. 如果目标目录不存在，则创建它们
    if not os.path.exists(low_dir):
        os.makedirs(low_dir)
        print(f"已创建文件夹: {low_dir}")
    
    if not os.path.exists(high_dir):
        os.makedirs(high_dir)
        print(f"已创建文件夹: {high_dir}")

    # 3. 读取 CSV 文件
    try:
        df = pd.read_csv(csv_path)
        print("成功读取 CSV 文件。")
    except FileNotFoundError:
        print(f"错误: 找不到 CSV 文件: {csv_path}")
        return

    # ================== 关键配置区 (请检查这里) ==================
    # 请将 'Filename' 修改为你 CSV 中存放文件名的那一列的表头名称
    # 如果 CSV 里的文件名不带 .svs 后缀，脚本会在下方自动尝试添加
    filename_col = 'ID'  # 例如可能是 'file_name', 'case_id', 'slide_id' 等
    
    # 确保 Label 列存在 (区分大小写，根据你的描述是 'Label')
    label_col = 'label'
    # ==========================================================

    # 检查列名是否存在
    if filename_col not in df.columns or label_col not in df.columns:
        print(f"错误: CSV 中未找到列名 '{filename_col}' 或 '{label_col}'。")
        print(f"检测到的列名有: {list(df.columns)}")
        return

    # 4. 遍历并移动文件
    moved_count = 0
    missing_count = 0

    for index, row in df.iterrows():
        # 获取文件名和标签
        fname = str(row[filename_col]).strip()
        label = row[label_col]

        # 如果 CSV 里的文件名没有 .svs 后缀，自动补全
        if not fname.endswith('.svs'):
            fname += '.svs'

        # 构建完整的源文件路径
        source_file = os.path.join(source_dir, fname)

        # 确定目标文件夹
        if label == 1:
            dest_folder = high_dir
        elif label == 0:
            dest_folder = low_dir
        else:
            print(f"跳过文件 {fname}: Label 数值 ({label}) 不是 0 或 1")
            continue

        dest_file = os.path.join(dest_folder, fname)

        # 执行移动操作
        if os.path.exists(source_file):
            try:
                # 使用 move 移动文件
                shutil.move(source_file, dest_file)
                print(f"[移动成功] {fname} -> {os.path.basename(dest_folder)}")
                moved_count += 1
            except Exception as e:
                print(f"[移动失败] {fname}: {e}")
        else:
            # 尝试查找是否存在该文件但在目标文件夹里（避免重复移动报错）
            if os.path.exists(dest_file):
                print(f"[已存在] {fname} 已经在目标文件夹中，跳过。")
            else:
                print(f"[文件缺失] 在源目录找不到文件: {fname}")
                missing_count += 1

    print("-" * 30)
    print(f"处理完成。共移动: {moved_count} 个文件。缺失/未找到: {missing_count} 个文件。")

if __name__ == "__main__":
    move_svs_files()
```

```python
import os
import shutil

# --- 路径设置 ---
# 包含 High 和 Low 的数据集根目录
dataset_root = r'G:\1.Sprouts\CRC.Pathology\20.WSL\Dataset'
# 存放已勾画 Patch 的源目录
patch_source_dir = r'F:\WSI.patches\TCGA_COAD'

# 需要处理的子文件夹名
sub_folders = ['High', 'Low']

def sync_roi_patches():
    count_success = 0
    count_missing = 0

    for sub in sub_folders:
        # 构建当前分类的完整路径 (G:\...\High)
        current_sub_path = os.path.join(dataset_root, sub)
        
        # 定义该分类下存放 ROI 文件夹的目标位置 (G:\...\High\COAD.ROI)
        target_roi_base = os.path.join(current_sub_path, 'COAD.ROI')
        
        # 如果目标 ROI 根目录不存在则创建
        if not os.path.exists(target_roi_base):
            os.makedirs(target_roi_base)
            print(f"已创建目标目录: {target_roi_base}")

        # 遍历当前子文件夹下的所有文件
        for filename in os.listdir(current_sub_path):
            # 只处理 .svs 文件
            if filename.lower().endswith('.svs'):
                # 提取 ID (例如 TCGA-5M-AAT6)
                sample_id = os.path.splitext(filename)[0]
                
                # 构建 F 盘中对应的源文件夹路径
                source_patch_folder = os.path.join(patch_source_dir, sample_id)
                
                # 构建 G 盘中对应的目标存放路径
                target_patch_folder = os.path.join(target_roi_base, sample_id)

                # 检查 F 盘是否存在该 ID 的文件夹
                if os.path.exists(source_patch_folder) and os.path.isdir(source_patch_folder):
                    # 检查目标位置是否已经存在，避免重复复制
                    if not os.path.exists(target_patch_folder):
                        print(f"正在复制 [{sub}]: {sample_id} ...")
                        try:
                            # 复制整个文件夹及其内容
                            shutil.copytree(source_patch_folder, target_patch_folder)
                            count_success += 1
                        except Exception as e:
                            print(f"复制 {sample_id} 时出错: {e}")
                    else:
                        print(f"跳过 (已存在): {sample_id}")
                else:
                    # 如果 F 盘里没有这个 ID 的文件夹
                    count_missing += 1
                    # print(f"未找到对应 Patch: {sample_id}")

    print("\n" + "="*40)
    print(f"任务完成！")
    print(f"成功迁移文件夹: {count_success} 个")
    print(f"未在源端找到匹配: {count_missing} 个")
    print("="*40)

if __name__ == "__main__":
    sync_roi_patches()
```
