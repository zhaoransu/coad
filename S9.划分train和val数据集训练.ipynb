```python
import os
import shutil
import random
from pathlib import Path
import time

def split_dataset_train_val():
    # ================= 配置区域 =================
    # 源目录 (存放所有归一化后的 png)
    source_dir = Path(r"G:\1.Sprouts\Low\Low.patches.train.normalized")
    
    # 目标目录 (脚本会自动创建)
    train_dir = source_dir / "train"
    val_dir = source_dir / "val"
    
    # 划分比例 (0.6 代表训练集占 60%)
    TRAIN_RATIO = 0.6
    
    # 随机种子 (保证每次运行结果复现，如果想要完全随机可设为 None)
    RANDOM_SEED = 42
    # ===========================================

    # 1. 检查源目录
    if not source_dir.exists():
        print(f"❌ 错误: 源目录不存在 -> {source_dir}")
        return

    # 2. 创建目标文件夹
    train_dir.mkdir(parents=True, exist_ok=True)
    val_dir.mkdir(parents=True, exist_ok=True)

    # 3. 获取所有图片文件
    # glob("*.png") 只会搜索当前目录，不会递归搜索子目录(比如train/val)，这正是我们想要的
    print(f"🔍 正在扫描文件: {source_dir}")
    all_files = list(source_dir.glob("*.png"))
    total_files = len(all_files)

    if total_files == 0:
        print("⚠️  源目录下没有找到 .png 文件 (或者文件已经被移动了)")
        return

    # 4. 打乱顺序
    print(f"🎲 正在随机打乱 {total_files} 张图片 (种子: {RANDOM_SEED})...")
    random.seed(RANDOM_SEED)
    random.shuffle(all_files)

    # 5. 计算切分点
    split_index = int(total_files * TRAIN_RATIO)
    
    train_files = all_files[:split_index]
    val_files = all_files[split_index:]

    print(f"📊 划分计划:")
    print(f"   - 训练集 (60%): {len(train_files)} 张 -> {train_dir.name}")
    print(f"   - 验证集 (40%): {len(val_files)} 张 -> {val_dir.name}")
    print("-" * 50)

    # 6. 执行移动操作
    start_time = time.time()
    
    def move_files(file_list, target_folder, label):
        count = 0
        for f in file_list:
            try:
                dest = target_folder / f.name
                shutil.move(str(f), str(dest))
                count += 1
                if count % 1000 == 0:
                    print(f"   [{label}] 已移动 {count} 张...")
            except Exception as e:
                print(f"   ❌ 移动失败 {f.name}: {e}")
        print(f"   ✅ [{label}] 完成。")

    print("🚀 开始移动文件...")
    move_files(train_files, train_dir, "Train")
    move_files(val_files, val_dir, "Val")

    duration = time.time() - start_time
    print("-" * 50)
    print(f"🎉 全部处理完毕！耗时: {duration:.2f} 秒")
    print(f"📁 最终结构:")
    print(f"   {train_dir}")
    print(f"   {val_dir}")

if __name__ == "__main__":
    split_dataset_train_val()
```
